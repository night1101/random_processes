```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

means = df.groupby("sample")["x"].mean()
print(means)

plt.figure(figsize=(12, 7))
for sample, group in df.groupby("sample"):
    plt.plot(group["t"], group["x"], marker="o", markersize=3, label=f"sample {sample}")

plt.xlabel("t")
plt.ylabel("x")
plt.title("Траектории марковской цепи")
plt.legend()
plt.grid(True)
plt.show()
```

```python
for sample, group in df.groupby("sample"):
    plt.figure(figsize=(8, 4))
    plt.plot(group["t"], group["x"], marker="o", markersize=3)
    plt.xlabel("t")
    plt.ylabel("x")
    plt.title(f"Траектория sample = {sample}")
    plt.grid(True)
    plt.show()
```
